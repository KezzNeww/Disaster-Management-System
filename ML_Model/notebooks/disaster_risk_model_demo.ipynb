## 5. Live Demonstration: Incident Inference
We can feed simulated real-time emergency sensor readings to test instant inference.

```python
predictor = DisasterRiskPredictor('../models/risk_model.pkl')

# Scenario 1: Extreme Flash Flood Event
flood_incident = {
    'hazard_type': 'Flood',
    'rainfall_mm_24h': 235.0,
    'river_discharge_m3s': 1850.0,
    'soil_moisture_pct': 96.0,
    'ambient_temp_c': 23.5,
    'wind_speed_kmh': 55.0,
    'humidity_pct': 95.0,
    'seismic_magnitude': 0.1,
    'fault_distance_km': 60.0,
    'slope_gradient_deg': 8.0,
    'vegetation_ndvi': 0.40,
    'population_density_sqkm': 5400.0,
    'infrastructure_vulnerability': 0.85,
    'elevation_m': 30.0,
    'distance_to_water_body_km': 0.8,
    'drainage_capacity_index': 0.20
}

pred_flood = predictor.predict_one(flood_incident)
print("Scenario 1 - Flash Flood:")
print(f"  Predicted Risk Level : {pred_flood['predicted_risk_level']}")
print(f"  Confidence           : {pred_flood['confidence']*100:.2f}%")
print(f"  Probabilities        : {pred_flood['class_probabilities']}")
```

```python
# Scenario 2: Moderate High-Wind / Wildfire Risk Event
wildfire_incident = {
    'hazard_type': 'Wildfire',
    'rainfall_mm_24h': 0.0,
    'river_discharge_m3s': 20.0,
    'soil_moisture_pct': 8.5,
    'ambient_temp_c': 41.0,
    'wind_speed_kmh': 65.0,
    'humidity_pct': 14.0,
    'seismic_magnitude': 0.0,
    'fault_distance_km': 90.0,
    'slope_gradient_deg': 18.0,
    'vegetation_ndvi': 0.58,
    'population_density_sqkm': 1200.0,
    'infrastructure_vulnerability': 0.60,
    'elevation_m': 450.0,
    'distance_to_water_body_km': 15.0,
    'drainage_capacity_index': 0.60
}

pred_fire = predictor.predict_one(wildfire_incident)
print("\nScenario 2 - Wildfire Threat:")
print(f"  Predicted Risk Level : {pred_fire['predicted_risk_level']}")
print(f"  Confidence           : {pred_fire['confidence']*100:.2f}%")
print(f"  Probabilities        : {pred_fire['class_probabilities']}")
```
